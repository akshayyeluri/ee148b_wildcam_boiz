```python
import numpy as np
import matplotlib.pyplot as plt
import os

prefix = "/home/ubuntu/data"
DATASET = prefix
MODEL_PATHS = ["./efficientnet_rescale_v1_aws", "./efficientnet_orig_ratio"]
labels = ["Resize Crop", "Pad Crop"]
```

```python
train_losses = []
train_accs = []
val_losses = []
val_accs = []
for MODEL_PATH in MODEL_PATHS:
    with open(os.path.join(MODEL_PATH, f"train_losses.npy"), 'rb') as f:
        train_losses.append(np.load(f))
    with open(os.path.join(MODEL_PATH, f"train_accs.npy"), 'rb') as f:
        train_accs.append(np.load(f))
    with open(os.path.join(MODEL_PATH, f"val_losses.npy"), 'rb') as f:
        val_losses.append(np.load(f))
    with open(os.path.join(MODEL_PATH, f"val_accs.npy"), 'rb') as f:
        val_accs.append(np.load(f))


plt.figure()
plt.loglog()
for val_loss in val_losses:
    plt.plot(val_loss)

plt.title("Efficientnet Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Cross Entropy Loss")
plt.legend(labels)
plt.show()

plt.figure()
plt.loglog()
for val_acc in val_accs:
    plt.plot(val_acc)
plt.title("Efficientnet Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend(labels)
plt.figure()
plt.show()
```
